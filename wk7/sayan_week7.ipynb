## Structure of the Westminster data

The JSON file loads as a Python dictionary. Its resources key
contains a list of 13 tables: the overall register and 12
category tables.

I selected the overall register. Its data key contains a list
of 4,070 entries. This counts declarations, not individual MPs,
because one MP can have several entries.

Each entry is a dictionary containing fields such as member,
party, category and summary.

For example, the first entry concerns Ed Davey. Its summary
lists National Liberal Club - £798.00, under the category
Gifts, benefits and hospitality from UK sources.

## Question 1

How many entries concern gifts, benefits and hospitality
from UK sources?

```python
# Count entries for gifts, benefits and hospitality from UK sources
gifts = 0

for entry in commons:
    if entry["category"] == "Gifts, benefits and hospitality from UK sources":
        gifts += 1

print(gifts)
```

There are 645 entries in the category “Gifts, benefits and
hospitality from UK sources” in the supplied overall register.

I checked the category of each entry and counted exact matches.
This counts entries, not individual MPs, because an MP can
have more than one declaration in this category.

## Question 2

How many different MPs have entries for gifts, benefits
and hospitality from UK sources?

```python
# Count each MP once using their parliamentary ID
gift_mps = {}

for entry in commons:
    if entry["category"] == "Gifts, benefits and hospitality from UK sources":
        gift_mps[entry["mnis_id"]] = entry["member"]

print(len(gift_mps))
```

The 645 entries for gifts, benefits and hospitality from UK
sources relate to 279 different MPs in the supplied file.

I used each MP’s parliamentary ID (mnis_id) as a dictionary key.
Each ID appears only once in the dictionary, so repeated
declarations by the same MP do not increase the MP count.

## UK gifts, benefits and hospitality entries by MP

The table below lists all 279 MPs with entries in this category
in the supplied file. It shows their names, parties and number
of entries, ordered from the highest count to the lowest.

I grouped entries using each MP’s parliamentary ID, so repeated
declarations contribute to that MP’s count.

These are counts of registered entries. One entry can cover
several gifts or benefits.

```python
import pandas as pd

# Count UK gifts, benefits and hospitality entries for each MP
gift_counts = {}

for entry in commons:
    if entry["category"] == "Gifts, benefits and hospitality from UK sources":
        mp_id = entry["mnis_id"]

        if mp_id not in gift_counts:
            gift_counts[mp_id] = {
                "MP": entry["member"],
                "Party": entry["party"],
                "UK gifts entries": 0
            }

        gift_counts[mp_id]["UK gifts entries"] += 1

# Create a table with one row per MP
gift_table = pd.DataFrame.from_dict(gift_counts, orient="index")
gift_table.index.name = "Parliamentary ID"

# Put MPs with the most entries first
gift_table = gift_table.sort_values("UK gifts entries", ascending=False)

# Display every row without truncating the table
with pd.option_context("display.max_rows", None):
    display(gift_table)
```

## Senedd data: structure and scope

The file contains one table called Entries, with 1,755 records.
Each record is a dictionary containing fields such as person_id,
person_name, category_name, entry_id and content.

The description gives March 2025 as the starting register date.
The first record concerns Paul Rock MS and describes complimentary
circus tickets and a VIP reception.

The in_latest_register field is True for 10 records and False
for 1,745. The analysis below uses all records in the supplied
file. The coverage of this status field needs checking before
drawing conclusions about current registered interests.

## Senedd question 1

Which category labels have the most entries in the supplied file?

```python
# Count records under each category label
category_counts = collections.defaultdict(int)

for entry in senedd_entries:
    category_counts[entry["category_name"]] += 1

# Rank category labels from the most entries to the fewest
category_counter = collections.Counter(category_counts)
category_counter.most_common()
```

The most frequent category label is “Membership/Chairmanship
of bodies in receipt of Senedd funds”, with 347 entries.

This is followed by “Standing Order 5A: Voluntary record of
Members’ other interests” with 316, and “Remunerated Employment,
Directorships etc.” with 181.

I counted category names exactly as they appear in the file.
Similar labels, including “Land and Property” and “Land and
property”, appear separately. Their wording and definitions
would need checking before combining them into broader categories.

## Senedd question 2

Which members have the most entries in the supplied file?

```python
# Count entries by member ID and keep names for display
member_counts = collections.defaultdict(int)
member_names = {}

for entry in senedd_entries:
    person_id = entry["person_id"]
    member_counts[person_id] += 1
    member_names[person_id] = entry["person_name"]

# Display the ten members with the most entries
member_counter = collections.Counter(member_counts)

for person_id, count in member_counter.most_common(10):
    print(member_names[person_id], "-", count)
```

The three members with the most entries in the supplied file are:

| Member | Entries |
|---|---:|
| Jayne Bryant | 85 |
| Darren Millar | 76 |
| Sioned Williams | 67 |

I counted entries using person_id to identify each member,
then used Counter.most_common(10) to display the ten highest counts.

The ranking includes all records in the supplied file, regardless
of their latest-register status. These are counts of entries
across all categories; conclusions about current interests
require the status-field checks described above.
