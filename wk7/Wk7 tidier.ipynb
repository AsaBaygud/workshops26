```python
print(json.dumps(d["resources"][6],indent=4))
```

```python
visits_abroad = d["resources"][6]["data"]
```

```python
len(visits_abroad)
```

```python
mp_visits = collections.defaultdict(int) 

for visits in visits_abroad:
    mp_visits[visits["member"]] += 1

print(mp_visits)
    
```

```python
mp_counter = collections.Counter(mp_visits)
mp_counter.most_common(10)
```

```python
mp_names = set()
for visit in visits_abroad:
    mp_names.add(visit["member"])

mp_names = list(mp_names)
print(mp_names)
```

```python
nepo_babies = d["resources"][11]["data"]
```

```python
len(nepo_babies)
```

```python
nepo_baby_party = collections.defaultdict(int) 

for entry in nepo_babies:
    nepo_baby_party[entry["party"]] += 1

print(nepo_baby_party)
```

```python
nepo_relations = collections.defaultdict(int) 

for entry in nepo_babies:
    nepo_relations[entry["family_relation_type"]] += 1

print(nepo_relations)
```

```python
nepo_members = set()

for entry in nepo_babies:
    nepo_members.add(entry["member"])

nepo_members = list(nepo_members)
print(nepo_members)
```
