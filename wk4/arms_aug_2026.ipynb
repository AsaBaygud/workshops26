### Filtering!

```python
# show a filtered version with just Austria
items[items.destination == 'Austria'].head(3)

# save the filtered df
aut = items[items.destination == 'Austria']

# export the csv (goes to the same directory as this notebook)
aut.to_csv("austria_df.csv", index = False)
```

```python
# check data type on cols - is the 'date' column really datetime? No

items.info()
```

```python
# make a new 'year' col from the date col
items["year"] = items["date"].str[:4]
items.tail(2)
```

```python
# filter for years

# list of years we want
years = ['2024', '2025', '2026']

# filter against list
items[items.year.isin(years)]

# assign to a variable
threeyears = items[items.year.isin(years)]

# how many rows now?
print(len(threeyears))
```
