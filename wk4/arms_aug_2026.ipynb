### The 'onward' problem: incorporation destinations

- Filter the incorporation-destinations csv for 'Israel' as the destination. There are 2663 cohort ids that include Israel
- Make a list of all the licence_cohort_id numbers. We'll call the list 'LCI'. There are 3573 (so not all these rows concern Israel it seems)
- Create a new column in the 'items' df called 'can include ISR' with 1s and 0s.

```python
# filter the incorporation-destinations df (inc) and print number of rows (len)

print(len(inc[inc.destination == 'Israel']))
inc[inc.destination == 'Israel'].head(4)
```

```python
# Example: show me just rows with licence_cohort_id 195753. This cohort id includes several countries. 

print(len(items[items.licence_cohort_id == 195753]))
items[items.licence_cohort_id == 195753].head(3)
```

```python
# in the INC dataset how many different countries appear in the LICENCE_COHORT_ID number 195753. 

# if you want the countries as a list
# inc[inc.licence_cohort_id == 195753].destination.tolist()

# otherwise
inc[inc.licence_cohort_id == 195753]
```

```python
# get all the cohort ids in INC in which Israel appears.

lci = list(inc[inc.destination == 'Israel'].licence_cohort_id)
lci
```

```python
# make new column that flags all the rows with the selected licence_cohort_ids

items["can_include_ISR"] = items["licence_cohort_id"].isin(lci).astype(int)
```

```python
items.head(4)
```

```python
# where israel is included as a possible final destination, what is the initial country?

items[items.can_include_ISR == 1].destination.value_counts()
```
